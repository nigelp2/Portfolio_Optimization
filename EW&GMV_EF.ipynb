```python
import edhec_risk_kit as erk
import numpy as np
import pandas as pd

ind = erk.get_ind_returns()
er = erk.annualize_rets(ind['1995':'2000'], 12)
cov = ind['1995':'2000'].cov()
```

```python
l = ['Food', 'Steel']
erk.msr(0.1, er[l], cov.loc[l,l])
```
